```python
df["Rainfall_mm"]= df["Rainfall_mm"].fillna(df["Rainfall_mm"].median())
```

```python
df["Soil_Moisture_pct"]= df["Soil_Moisture_pct"].fillna(df["Soil_Moisture_pct"].median())
```

```python
df.dropna(inplace=True)
```

```python
numerical_columns = df.select_dtypes(include='number').columns

for col in numerical_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    outliers = df[(df[col] < lower) | (df[col] > upper)]
    
    print(col, ":", len(outliers), "outliers")

    print(f"{col}")
    print(f"Lower Bound: {lower}")
    print(f"Upper Bound: {upper}")
```

How does agricultural performance vary across seasons?

```python


sns.barplot(
    data=df,
    x='Season',
    y='Production_Tonnes'
)

plt.title('Average Crop Production Across Seasons')
plt.xlabel('Season')
plt.ylabel('Production (Tonnes)')
plt.show()
```

What major seasonal patterns can be observed?

```python
season_yield = df.groupby('Season')['Yield_Tonnes_Ha'].mean().reset_index()

sns.lineplot(
    data=season_yield,
    x='Season',
    y='Yield_Tonnes_Ha',
    marker='o'
)

plt.title('Seasonal Pattern in Average Crop Yield')
plt.xlabel('Season')
plt.ylabel('Average Yield (Tonnes/Ha)')
plt.show()
```

Which characteristics change between seasons?

```python
season_features = df.groupby('Season')[
    ['Rainfall_mm',
     'Avg_Temperature_C',
     'Humidity_pct',
     'Soil_Moisture_pct',
     'Yield_Tonnes_Ha']
].mean()

sns.heatmap(season_features, annot=True, fmt='.2f', cmap='YlGnBu')

plt.title('Agricultural Characteristics Across Seasons')
plt.xlabel('Characteristics')
plt.ylabel('Season')
plt.show()
```

What differences exist between agricultural activities in different seasons?

```python
sns.countplot(
    data=df,
    x='Season',
    hue='Crop'
)

plt.title('Crop Distribution Across Seasons')
plt.xlabel('Season')
plt.ylabel('Number of Farms')
plt.show()
```

Are there noticeable variations in resource usage across seasons?

```python
sns.barplot(
    data=df,
    x='Season',
    y='Water_Used_m3'
)

plt.title('Average Water Usage Across Seasons')
plt.xlabel('Season')
plt.ylabel('Average Water Used (m³)')
plt.show()
```

Are there relationships between seasonal environmental conditions and agricultural performance?

```python
sns.scatterplot(
    data=df,
    x='Rainfall_mm',
    y='Yield_Tonnes_Ha',
    hue='Season'
)

plt.title('Rainfall vs Crop Yield Across Seasons')
plt.xlabel('Rainfall (mm)')
plt.ylabel('Yield (Tonnes/Ha)')
plt.show()
```

How do economic outcomes vary across seasons?

```python
sns.barplot(
    data=df,
    x='Season',
    y='Profit_INR'
)

plt.title('Average Profit Across Seasons')
plt.xlabel('Season')
plt.ylabel('Average Profit (INR)')
plt.show()
```

Are some seasonal patterns consistent across different regions or categories?

```python
sns.barplot(
    data=df,
    x='State',
    y='Yield_Tonnes_Ha',
    hue='Season'
)

plt.title('Average Yield Across States and Seasons')
plt.xlabel('State')
plt.ylabel('Average Yield (Tonnes/Ha)')
plt.xticks(rotation=45)
plt.show()
```

Are there unusual or unexpected seasonal patterns?

```python
sns.boxplot(
    data=df,
    x='Season',
    y='Disease_Pest_Risk_pct'
)

plt.title('Disease and Pest Risk Across Seasons')
plt.xlabel('Season')
plt.ylabel('Disease/Pest Risk (%)')
plt.show()
```

What insights can be derived from the observed seasonal differences?

```python
insight_data = df.groupby('Season')[
    ['Yield_Tonnes_Ha',
     'Profit_INR',
     'Water_Used_m3',
     'Rainfall_mm',
     'Disease_Pest_Risk_pct']
].mean()

sns.heatmap(insight_data, annot=True, fmt='.2f', cmap='coolwarm')

plt.title('Overall Seasonal Agricultural Performance')
plt.show()
```
