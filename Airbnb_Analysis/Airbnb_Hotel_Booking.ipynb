```python
# Handling null Values of Categorical Columns 

categorical_cols = ['host_identity_verified','instant_bookable','country','country code'
                    ,'neighbourhood group','neighbourhood','cancellation_policy']

for col in categorical_cols:
    df[col]= df[col].fillna(df[col].mode()[0])
```

```python
df.isnull().sum()
```

```python
# Handling missing values like names and host name 
df['NAME']=df['NAME'].fillna("Unknown")
df['host name']=df['host name'].fillna("Unknown")
```

```python
# handling missing values of numerical columns
numerical_cols = ['lat','long', 'Construction year','minimum nights','number of reviews','review rate number',
                    'calculated host listings count','availability 365']

for col in numerical_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR
    
    outliers_count = ((df[col] < lower_limit) | 
                      (df[col] > upper_limit)).sum()
    
    print(col, ":", outliers_count, "outliers")
```

```python
numerical_cols = ['lat','long', 'Construction year','minimum nights','number of reviews','review rate number',
                    'calculated host listings count','availability 365']

for col in numerical_cols:
    
    # Calculate Q1 and Q3
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    
    # Calculate IQR
    IQR = Q3 - Q1
    
    # Calculate limits
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR
    
    # Get outlier values
    outliers = df[
        (df[col] < lower_limit) | 
        (df[col] > upper_limit)
    ][col]
    
    # Print only columns that have outliers
    if len(outliers) > 0:
        print("\nColumn:", col)
        print("Number of outliers:", len(outliers))
        print("Lower limit:", lower_limit)
        print("Upper limit:", upper_limit)
        print("Outlier values:")
        print(outliers.values)
```

```python
numerical_cols = ['lat','long','Construction year','price','service fee','minimum nights','number of reviews',
                  'review rate number','calculated host listings count','availability 365']

for col in numerical_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
df['house_rules'] = df['house_rules'].fillna('Not Available')
```

```python
df['reviews per month'] = df['reviews per month'].fillna(0)

df['last review'] = df['last review'].fillna('No Review')
```

```python
df['price'] = df['price'].replace('[\$,]', '', regex=True)
df['price'] = pd.to_numeric(df['price'], errors='coerce')


df['service fee'] = df['service fee'].replace('[\$,]', '', regex=True)
df['service fee'] = pd.to_numeric(df['service fee'], errors='coerce')


df['price'] = df['price'].fillna(df['price'].median())

df['service fee'] = df['service fee'].fillna(
    df['service fee'].median())
```

```python
df.drop(columns='license', inplace=True)
```

```python
avg_price_room = (
    df.groupby('room type')['price']
    .mean()
    .sort_values(ascending=False)
)

print(avg_price_room)


plt.figure(figsize=(8, 5))

sns.barplot(
    x=avg_price_room.index,
    y=avg_price_room.values
)

plt.title("Average Price by Room Type")
plt.xlabel("Room Type")
plt.ylabel("Average Price")

plt.show()
```

```python
avg_price_location = (
    df.groupby('neighbourhood group')['price']
    .mean()
    .sort_values(ascending=False)
)

print(avg_price_location)

plt.figure(figsize=(10, 5))

sns.barplot(
    x=avg_price_location.index,
    y=avg_price_location.values
)

plt.title("Average Airbnb Price by Neighbourhood Group")
plt.xlabel("Neighbourhood Group")
plt.ylabel("Average Price")

plt.xticks(rotation=45)
plt.show()
```

```python
avg_reviews_room = (
    df.groupby('room type')['number of reviews']
    .mean()
    .sort_values(ascending=False)
)

print(avg_reviews_room)

plt.figure(figsize=(8, 5))

sns.barplot(
    x=avg_reviews_room.index,
    y=avg_reviews_room.values
)

plt.title("Average Number of Reviews by Room Type")
plt.xlabel("Room Type")
plt.ylabel("Average Number of Reviews")

plt.xticks(rotation=45)
plt.show()
```

```python
demand = (
    df.groupby('room type')['reviews per month']
    .mean()
    .sort_values(ascending=False)
)

print(demand)

plt.figure(figsize=(8,5))

sns.barplot(
    x=demand.index,
    y=demand.values
)

plt.title("Guest Demand by Room Type")
plt.xlabel("Room Type")
plt.ylabel("Average Reviews Per Month")

plt.show()
```

```python
host_reviews = (
    df.groupby('host name')['number of reviews']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(host_reviews)
plt.figure(figsize=(10, 6))

sns.barplot(
    x=host_reviews.values,
    y=host_reviews.index
)

plt.title("Top 10 Hosts by Number of Reviews")
plt.xlabel("Total Number of Reviews")
plt.ylabel("Host Name")

plt.show()
```

```python
availability_room = (
    df.groupby('room type')['availability 365']
    .mean()
    .sort_values(ascending=False)
)

print(availability_room)
plt.figure(figsize=(8, 5))

sns.barplot(
    x=availability_room.index,
    y=availability_room.values
)

plt.title("Average Listing Availability by Room Type")
plt.xlabel("Room Type")
plt.ylabel("Average Availability (Days)")

plt.xticks(rotation=45)
plt.show()
```

```python
location_performance = (
    df.groupby('neighbourhood group')['number of reviews']
    .mean()
    .sort_values(ascending=False)
)

print(location_performance)

plt.figure(figsize=(10, 5))

sns.barplot(
    x=location_performance.index,
    y=location_performance.values
)

plt.title("Location Performance by Neighbourhood Group")
plt.xlabel("Neighbourhood Group")
plt.ylabel("Average Number of Reviews")

plt.xticks(rotation=45)
plt.show()
```
